```python
import socket
```

```python
def getBfromUDP(port_no):
    # Set up the server details
    UDP_IP = socket.gethostname()  # Replace with the server's IP address
    UDP_PORT = port_no   # Replace with the server's port number
    MESSAGE = "STATUS"   # The message to send
    # STATUS requests the magnet field and slew rate

    # Create a UDP socket
    sock = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)

    try:
        # Send the message to the server
        sock.sendto(bytes(MESSAGE, 'UTF-8'), (UDP_IP, UDP_PORT))
        # print(f"Sent to server: {MESSAGE}")

        # Receive response from the server (buffer size is 1024 bytes)
        data, server = sock.recvfrom(1024)
        decoded_data = data.decode()
        # print(f"Received from server: {decoded_data}")

    finally:
        # Close the socket
        sock.close()

    # return current field and slew rate reading
    return decoded_data.split('\t')
```

```python
def set_B(B):
        UDP_IP = socket.gethostname()  # Replace with the server's IP address
        UDP_PORT = UDP_PORT            # Replace with the server's port number
        MESSAGE = f"SET SETPOINT {B}"   # The message to send
        # SET SETPOINT {B} writes the desired field value to the magnet

        # Create a UDP socket
        sock = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)

        try:
            # Send the message to the server
            sock.sendto(bytes(MESSAGE, 'UTF-8'), (UDP_IP, UDP_PORT))
        finally:
            # Close the socket
            sock.close()
```
